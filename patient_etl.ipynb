## Create Sample Healthcare Data

```sql
%python
patient_data = [
    (101, "John", 45, "Diabetes"),
    (102, "Mary", 32, "Asthma"),
    (103, "David", 65, "Hypertension"),
    (104, "Sarah", 29, "Asthma")
]

columns = [
    "patient_id",
    "patient_name",
    "age",
    "diagnosis"
]

df = spark.createDataFrame(patient_data, columns)

display(df)
```

## Add New Business Requirement

```sql
%python
# Healthcare team requests: Flag all patients above age 60 as Senior Citizens

# added senior patient logic
from pyspark.sql.functions import when, col

df_transformed = (
    df.withColumn(
        "senior_patient_flag",
        when(col("age") > 60, "Y")
        .otherwise("N")
    )
)

display(df_transformed)
```

## Add Risk Categorization

```sql
%python

from pyspark.sql.functions import when, col

df_risk = (
    df.withColumn(
        "risk_category",
        when(col("age") > 60, "HIGH")
        .when(col("age") >= 40, "MEDIUM")
        .otherwise("LOW")
    )
)

display(df_risk)

```
